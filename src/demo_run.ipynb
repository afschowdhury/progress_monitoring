```python

# %% [markdown]
# ## Method 2: Creating Analyzer Instance
# 
# For more control, you can create an analyzer instance directly.

# %% Create analyzer with custom configuration
if PROVIDER == "gemini":
    analyzer = pm.create_gemini_analyzer(
        api_key=API_KEY,
        model_name="gemini-2.0-flash-exp",
        memory_file="demo_memory.txt"
    )
elif PROVIDER == "openai":
    analyzer = pm.create_openai_analyzer(
        api_key=API_KEY,
        model_name="gpt-4o",
        memory_file="demo_memory.txt"
    )

print(f"Created {PROVIDER} analyzer")

# %% Run analysis with custom analyzer
try:
    results = analyzer.analyze_construction_site(IMAGES_FOLDER)
    
    # Display detailed results
    print("\n=== Analysis Results ===")
    print(f"Date: {results.get('date')}")
    print(f"Images analyzed: {results.get('images_analyzed', 0)}")
    print(f"Overall progress: {results.get('overall_progress', 'N/A')}")
    
    if 'detailed_analysis' in results:
        print(f"\nDetailed Analysis:\n{results['detailed_analysis']}")
    
    if 'progress_summary' in results:
        print(f"\nProgress Summary:\n{results['progress_summary']}")
        
except Exception as e:
    print(f"Analysis failed: {e}")

# %% [markdown]
# ## Method 3: Using Custom Prompts
# 
# You can customize the analysis by providing your own prompts.

# %% Create analyzer with custom prompts
custom_prompts = {
    "construction_analysis": """
    Analyze these construction site images and provide:
    1. Current phase of construction
    2. Visible progress since last analysis
    3. Quality observations
    4. Safety compliance notes
    5. Weather impact assessment
    """,
    "progress_assessment": """
    Based on the analysis, provide a progress assessment including:
    - Percentage completion estimate
    - Timeline adherence
    - Potential delays or concerns
    - Recommendations for next steps
    """
}

analyzer_with_prompts = pm.create_analyzer_with_custom_prompts(
    provider=PROVIDER,
    api_key=API_KEY,
    prompts_dir=None,  # We're using custom_prompts instead
    custom_prompts=custom_prompts,
    memory_file="demo_custom_memory.txt"
)

print("Created analyzer with custom prompts")

# %% Run analysis with custom prompts
try:
    custom_results = analyzer_with_prompts.analyze_construction_site(IMAGES_FOLDER)
    
    print("\n=== Custom Prompt Analysis Results ===")
    print(json.dumps(custom_results, indent=2, default=str))
    
except Exception as e:
    print(f"Custom analysis failed: {e}")

# %% [markdown]
# ## Viewing Memory and Progress History
# 
# The system maintains memory of previous analyses for progress tracking.

# %% Check memory contents
from progress_monitoring import MemoryManager

memory_manager = MemoryManager("demo_memory.txt")
memory_data = memory_manager.read_memory()

print("=== Memory Contents ===")
print(f"Project start date: {memory_data.get('project_start_date', 'Not set')}")
print(f"Total analysis days: {len(memory_data.get('daily_reports', {}))}")

if memory_data.get('daily_reports'):
    print("\nDaily reports:")
    for date, report in memory_data['daily_reports'].items():
        print(f"  {date}: {report.get('images_count', 0)} images analyzed")

# %% [markdown]
# ## Utility Functions
# 
# Some helpful utilities for working with the results.

# %% Save results to file
def save_results(results, filename):
    """Save analysis results to a JSON file."""
    with open(filename, 'w') as f:
        json.dump(results, f, indent=2, default=str)
    print(f"Results saved to {filename}")

# Save the latest results
if 'results' in locals():
    save_results(results, "latest_analysis.json")

# %% Display configuration info
print("=== Configuration Summary ===")
print(f"Provider: {PROVIDER}")
print(f"Available models:")
if PROVIDER == "gemini":
    print("  - gemini-2.0-flash-exp (recommended)")
    print("  - gemini-1.5-pro")
    print("  - gemini-1.5-flash")
elif PROVIDER == "openai":
    print("  - gpt-4o (recommended)")
    print("  - gpt-4-turbo")
    print("  - gpt-4")

print(f"\nPackage version: {pm.__version__}")
print("Available components:", ", ".join(pm.__all__))

# %% [markdown]
# ## Error Handling and Troubleshooting
# 
# Common issues and solutions.

# %% Test basic functionality
def test_setup():
    """Test if the setup is working correctly."""
    issues = []
    
    # Check API key
    if API_KEY == "your-api-key-here":
        issues.append("API key not set")
    
    # Check images folder
    if not Path(IMAGES_FOLDER).exists():
        issues.append(f"Images folder '{IMAGES_FOLDER}' does not exist")
    elif not list(Path(IMAGES_FOLDER).glob("*.jpg")) and not list(Path(IMAGES_FOLDER).glob("*.png")):
        issues.append(f"No image files found in '{IMAGES_FOLDER}'")
    
    if issues:
        print("⚠️  Setup Issues Found:")
        for issue in issues:
            print(f"   - {issue}")
        print("\nPlease fix these issues before running the analysis.")
    else:
        print("✅ Setup looks good!")
    
    return len(issues) == 0

test_setup()
```
